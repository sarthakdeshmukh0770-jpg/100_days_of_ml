```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
x,y=load_diabetes(return_X_y=True)
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
class SGDregresion:
    def __init__(self,lr=0.01,epoch=100):
        self.coff_=None
        self.intercept_=None
        self.lr=lr
        self.epoch=epoch
    def fit(self,x_train,y_train):
        self.intercept_=0
        self.coff_=np.ones(x_train.shape[1])
        for i in range(self.epoch):
            for j in range(x_train.shape[0]):
        
                idx=np.random.randint(0,x_train.shape[0])
                y_pred=np.dot(x_train[idx],self.coff_) + self.intercept_
                intercept_der=-2*(y_train[idx]-y_pred)
                self.intercept_=self.intercept_ - (self.lr * intercept_der)

                coff_dev=-2*np.dot((y_train[idx]-y_pred),x_train[idx])
                self.coff_=self.coff_ - (self.lr * coff_dev)
        
        print(self.intercept_,self.coff_)
    def predict(self,x_test):
        return np.dot(x_test,self.coff_) + self.intercept_
```

```python
lr=SGDregresion(epoch=50)
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```
