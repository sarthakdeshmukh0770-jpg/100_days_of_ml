# working with dates


```python
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
#extracting the year column
date['year']=date['date'].dt.year
```

```python
date.head()
```

```python
#extracting the months column 
date['months']=date['date'].dt.month
```

```python
date.head()
```

```python
# extreacting months names 
date['months_name']=date['date'].dt.month_name()
```

```python
date.head()
```

```python
# there are many functioned like thise
#dt.day
#dt.day0fweek
#dt.weak
#dt.quarter
```

```python
date['week_day_name']=date['date'].dt.day_name()
date.head()
```

```python
#is weekend?
date['is_weekend']=np.where(date['week_day_name'].isin(['Sunday','Saturday']),1,0)
```
